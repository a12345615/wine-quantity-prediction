data preprocessing

```python
#separate data and label
x=wine.drop('quality',axis=1)
```

```python
print(x)
```

label binary

```python
y=wine['quality'].apply(lambda y_value:1 if y_value >=7 else 0)
```

```python
y
```

```python
#train and test split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=2)
```

```python
print(x.shape,x_train.shape,x_test.shape)
```

```python
print(y.shape,y_train.shape,y_test.shape)
```

model training
random forest tree

```python
model=RandomForestClassifier()
```

```python
model.fit(x_train,y_train)
```

Evalution of the model

Accuracy score

```python
#accuracy on test data
x_test_prediction = model.predict(x_test)
test_data_accuracy=accuracy_score(x_test_prediction,y_test)
```

```python
print('accuracy_score:',test_data_accuracy)
```

build a predictive system

```python
input_data =(7.4,0.7,0.0,1.9,0.076,11.0,34.0,0.9978,3.51,0.56,9.4)
#changing the input data to a numpy array
input_data_as_numpy_array=np.asarray(input_data)
#reshape the numpy array the prediction the label for one value
input_data_reshaped=input_data_as_numpy_array.reshape(1,-1)

prediction = model.predict(input_data_reshaped)
print(prediction)

if (prediction[0]==1):
  print("good wine")
else:
  print("bad wine")
```
